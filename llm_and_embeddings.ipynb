```python
from transformers import pipeline, AutoTokenizer
import gensim.downloader as api
import numpy as np
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
generator = pipeline("text-generation", model="distilgpt2")
```

```python
prompt = "AI is transforming industries by"

outputs = generator(
    prompt,
    max_length=40,
    num_return_sequences=3
)

for i, output in enumerate(outputs):
    print(f"\nOutput {i+1}:")
    print(output['generated_text'])
```

**2. Tokenisation Demo**

```python
tokenizer = AutoTokenizer.from_pretrained("distilgpt2")

sentence = "LLMs are powerful tools for natural language understanding."

tokens = tokenizer.tokenize(sentence)
token_ids = tokenizer.convert_tokens_to_ids(tokens)

print("Tokens:")
print(tokens)

print("\nToken IDs:")
print(token_ids)

print("\nSequence Length:")
print(len(token_ids))
```

# Section C: Embeddings with Gensim

**4. Word Embeddings**

```python
model = api.load("glove-wiki-gigaword-50")
```

```python
words = ["king", "queen", "diamond"]

for word in words:

    print(f"\nWord: {word}")

    vector = model[word]

    print("First 10 vector values:")
    print(vector[:10])

    print("\nTop 5 similar words:")
    print(model.most_similar(word, topn=5))
```

# 5. Sentence-Level Embeddings

```python
sentences = [
    "Artificial intelligence is transforming technology",
    "Machine learning helps computers learn patterns",
    "Neural networks power modern AI systems",
    "Diamonds are valuable gemstones used in jewellery",
    "Jewellery often includes gold and diamonds"
]
```

```python
def sentence_vector(sentence):

    words = sentence.lower().split()

    vectors = [model[w] for w in words if w in model]

    return np.mean(vectors, axis=0)

sentence_vectors = np.array([sentence_vector(s) for s in sentences])
```

```python
similarity_matrix = cosine_similarity(sentence_vectors)

df_similarity = pd.DataFrame(
    similarity_matrix,
    index=sentences,
    columns=sentences
)

df_similarity
```

# Section D: Transformer Application

**Sentiment Analysis Example**

```python
sentiment = pipeline("sentiment-analysis")
```

```python
text1 = "The new AI system has greatly improved our customer service."

text2 = "The software update caused several unexpected problems."

print(sentiment(text1))
print(sentiment(text2))
```

Sentiment analysis can be used by businesses to analyse customer feedback from reviews, surveys, and social media posts. By automatically classifying large amounts of text data as positive or negative, organisations can quickly identify trends in customer satisfaction. This allows companies to detect potential issues early and respond more effectively to customer concerns. For example, an e-commerce company could analyse product reviews to understand customer sentiment toward specific products. Automating this process using transformer models reduces manual effort and helps organisations make faster data-driven decisions.
